# BoxPlot  ( Numeric vs Categorial) 
## Eg  ( Cholestrol vs HeartDisease)

```python
sns.boxplot(x = 'HeartDisease', y = 'Cholesterol', data = data)
```

### Cholesterol by heart-disease label

The two cholesterol distributions overlap. This chart describes the sample; it does not establish that cholesterol caused an outcome or estimate an individual's risk.

```python
sns.violinplot(x = 'HeartDisease', y = 'Age', data = data)
```

### Age by heart-disease label

The age distributions overlap, with older ages more represented in the positive-label group in this sample. This is a descriptive association, not evidence of causation.

```python
sns.heatmap(data.corr(  numeric_only  = True), annot = True)
```

### Reading the correlation heatmap

The heatmap summarizes pairwise linear correlations. A weak linear correlation does not rule out nonlinear or combined relationships, and correlation alone does not establish predictive value. Model-based evaluation would be needed to assess that.

# Data PreProcessing and Data Cleaning

```python
data_encode = pd.get_dummies(data, drop_first = True)
data_encode.head()
```

```python
boolean_cols = data_encode.select_dtypes(include="bool").columns
data_encode[boolean_cols] = data_encode[boolean_cols].astype(int)  # convert one-hot flags only
data_encode.head()
```

```python
from sklearn.preprocessing import StandardScaler
numeric_col = ['Age', 'RestingBP', 'Cholesterol', 'MaxHR', 'Oldpeak']
scaler = StandardScaler()
data_encode[numeric_col] = scaler.fit_transform(data_encode[numeric_col])
data_encode.head()
```
